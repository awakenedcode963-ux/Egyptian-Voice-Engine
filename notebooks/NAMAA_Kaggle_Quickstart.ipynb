```python
from pathlib import Path\nimport torchaudio as ta\nfrom huggingface_hub import snapshot_download\nfrom safetensors.torch import load_file\nfrom chatterbox import mtl_tts\nrepo_id = 'NAMAA-Space/NAMAA-Egyptian-TTS'\nmodel_dir = Path(snapshot_download(repo_id=repo_id, repo_type='model', revision='main'))\nmodel = mtl_tts.ChatterboxMultilingualTTS.from_pretrained(device=device)\ncheckpoint = model_dir / 't3_mtl23ls_v2.safetensors'\nmodel.t3.load_state_dict(load_file(str(checkpoint), device=device), strict=True)\nmodel.t3.to(device).eval()\nprint('NAMAA model loaded')\n
```
