## Permission gate
In a private working copy, record the consent record ID and approval only after the speaker explicitly agrees to synthetic and commercial use. Never commit that private data.


```python
import json
from pathlib import Path
from egyptian_voice_engine.catalog import VoiceProfile
from egyptian_voice_engine.engine import EgyptianVoiceEngine, GenerationConfig

profile_path = Path('voices/voice-001-noura-cairo-feminine/profile.json')
data = json.loads(profile_path.read_text(encoding='utf-8'))
data['reference']['local_path'] = REFERENCE_AUDIO
data['reference']['consent_status'] = 'approved'
data['reference']['consent_record_id'] = 'PRIVATE-CONSENT-ID'
profile_path.write_text(json.dumps(data, ensure_ascii=False, indent=2), encoding='utf-8')
profile = VoiceProfile.load(profile_path)
engine = EgyptianVoiceEngine(profile, GenerationConfig(device='cuda'))
result = engine.generate('أهلا بيكي، خلينا نبدأ خطوة خطوة', REFERENCE_AUDIO, 'outputs/noura_first_test.wav')
print(result)
```
